```python
asia = df[df['origin'] == 'Asia']
subset = asia[['vehicle_weight', 'model_year']].head(7)
subset
```

```python
X = subset.values
X
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
w.sum()
```
